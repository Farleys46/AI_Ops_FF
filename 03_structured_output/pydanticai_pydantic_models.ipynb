## Force a stuctured output using pydantic

```python
from pydantic import BaseModel, Field

class EmployeeModel(BaseModel):
    name: str = Field(description="för och efternamn, inte jättelångt, svenska eller internationella namn")
    age: int = Field(description="ålder från 18 till 67år", ge = 18, le = 67)
    salary: int = Field(description="lön mellan 30-60k,, ju mer senior position, desto högre lön",
                        ge=30000,
                        lt=60000,
    )
    role: str = Field(description="yrke/roll")
    description: str = Field(description="Kort förklaring, max 2 meningar på rollen")
employee1 = EmployeeModel(name="Bertil", age=40, salary=30000, role="IT-Supporter", description="Hjälper folk med IT problem")
employee1
```

```python
employee_simulator = Agent(
    model=nvidia_model("nvidia/nemotron-3.5-super-vl-120b-a12b"),
    instructions="Du är expert på att simulera anställda i Sverige",
)

result = await employee_simulator.run("Ge mig en data engineer i stockholm")

result.output
```

```python
result = await employee_simulator.run(
    "Ge mig en data engineer i stockholm", output_type=EmployeeModel
)
result
```
