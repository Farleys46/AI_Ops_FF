# Intro to LLMs and pydanticAI

```python
from pydantic_ai import Agent
from pydantic_ai.models.openai import OpenAIChatModel
from pydantic_ai.providers.openai import OpenAIProvider
import os
from dotenv import load_dotenv

load_dotenv()

model = OpenAIChatModel(
    model_name="nvidia/nemotron-3.5-super-vl-120b-a12b",
    provider=OpenAIProvider(
        base_url = "https://integrate.api.nvidia.com/v1",
        api_key = os.getenv("NVIDIA_API")
    ),
)

model
```

```python
agent = Agent(model=model, system_prompt="You are a helpful assistant that answers questions about Formula 1. Answer like a Formula 1 expert. If you don't know the answer, say 'David Coulthard would know this but i dont'. Answer with Emojis and other fun stuff.")

prompt = "How many times has Sebastian Vettel crashed in F1?"
result = await agent.run(prompt, timeout = 60)
result
```

```python
print(result.output)
```

```python
result.all_messages()
```

## Chefbot

```python
chef_agent = Agent(
    model=model,
    instructions=("You are a professional chef. You are a master of all cuisines and can answer any question about food, cooking, and recipes in a pedagogical manner. You are also a great conversationalist and can engage in friendly banter with the user."
                """Answer in the following format:
                Description 1-2 sentences
                List of ingredients in a python list
                Short step-by-step instructions how to make the dish"""
                ),
    
)

result = await chef_agent.run("Give me 3 Asian soups. Whole output should be in a python list format", timeout = 60)
print(result.output)
```

## Thinking Level

```python
from pydantic_ai import ModelSettings

storyteller = Agent(
    model=model,
    instructions=(
        "You are a professional storyteller. You are a master of all genres and can answer any question about storytelling, writing, and literature in a pedagogical manner. You are also a great conversationalist and can engage in friendly banter with the user."
        """Answer in the following format:
        Description max 3 sentences"""
    ),
    model_settings=ModelSettings(thinking="low", temperature=0)
)

prompt="A rabbit called bella likes to program in python ..."

story1 = await storyteller.run(prompt, timeout = 6000)
print(story1.output)
```

```python
story2 = await storyteller.run(prompt, timeout = 600)
print(story2.output)
```

## Multimodal

```python
from pathlib import Path
from pydantic_ai.messages import BinaryContent

image_joker = Agent(
    model=model,
    instructions="You are a person that can analyze images and answer questions about them. You are a master of all genres and can answer any question about images, photography, and art in a pedagogical manner. You are also a great conversationalist and can engage in friendly banter with the user.",
)

image_data = Path("bella.png").read_bytes()
prompt = "Tell me about this image and make a joke about it. Answer in a fun and humorous way."


result = await image_joker.run(
    user_prompt=[prompt, BinaryContent(data=image_data, media_type="image/png")]
)
print(result.output)
```
