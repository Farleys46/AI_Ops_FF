# PydanticAI with Nvidia Build

```python
from pydantic_ai import Agent
from pydantic_ai.models.openai import OpenAIChatModel
from pydantic_ai.providers.openai import OpenAIProvider
import os
from dotenv import load_dotenv

load_dotenv()

model = OpenAIChatModel(
    model_name="nvidia/nemotron-3.5-lightning-30b-a3b",
    provider=OpenAIProvider(
        base_url = "https://integrate.api.nvidia.com/v1",
        api_key = os.getenv("NVIDIA_API")
    ),
)

model
```

```python
agent = Agent(model)
result = await agent.run("Who is the best F1 Driver of all time?", timeout = 60)
result
```

```python
print(result.output)
```

```python
model = OpenAIChatModel(
    model_name="nvidia/nemotron-3.5-super-vl-120b-a12b",
    provider=OpenAIProvider(
        base_url = "https://integrate.api.nvidia.com/v1",
        api_key = os.getenv("NVIDIA_API")
    ),
)

result = await agent.run("How many wins in F1 does Kimi Raikkonen have?", timeout = 60, model = model)
result
```
