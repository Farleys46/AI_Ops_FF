### CREATE TABLE

```python
import json
with open ("data/animals_text_embeddings.json") as f:
    data = json.load(f)


# This data has vector embeddings already created
# for this it is generated by AI, but in reality if yiu use other vector databases that
# Dont automatically create embeddings you need to create the embeddings yourself
data
```

```python
data[0].get("text")
```

```python
data[0].get("vector")
```

```python
db.create_table("animals_text", exist_ok = True, data=data)
db["animals_text"]
```
